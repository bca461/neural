```python
def mcp_neuron(x1,x2,threshold):
    weighted_sum=x1+x2
    if weighted_sum>=threshold:
        return 1
    else:
        return 0
threshold=1
print("x1 x2 Output ")
inputs=[(0,0),
        (0,1),
        (1,0),
        (1,1)]
for x1,x2 in inputs:
    output=mcp_neuron(x1,x2,threshold)
    print(x1,"",x2,"",output)
    
```
