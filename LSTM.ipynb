```python
#implement sequence prediction using an LSTM (Long Short-Term Memory) neural network.*

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# Training data
X = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50],
    [40, 50, 60],
    [50, 60, 70]
])

y = np.array([40, 50, 60, 70, 80])

# Reshape data for LSTM
# (samples, time steps, features)
X = X.reshape((X.shape[0], X.shape[1], 1))

# Create LSTM model
model = Sequential([
    LSTM(50, activation='tanh', input_shape=(3, 1)),
    Dense(1)
])

# Compile model
model.compile(optimizer='adam', loss='mse')

# Train model
model.fit(X, y, epochs=500, verbose=0)

# Test input
test = np.array([[60, 70, 80]])
test = test.reshape((1, 3, 1))

# Prediction
prediction = model.predict(test, verbose=0)

print("Input sequence:", [60, 70, 80])
print("Predicted next value:", prediction[0][0])
```
